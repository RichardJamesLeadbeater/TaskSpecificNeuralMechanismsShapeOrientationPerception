```python
import itertools as itls
from IPython.display import display, Markdown
import os
import pandas as pd
import scipy.stats as st
from mytools.analysis import get_mean_data, get_ratio_data
from mytools.dframe_tools import init_cols, to_csv_pkl, index_dframe, custom_sort, spss_format
from mytools.dict_tools import append_dicty
from mytools.ratios import geomean
```

```python
"""
Initial settings common to both experiments
"""
# Paths #
og_dir = os.getcwd()

# Analysis settings #
participants = [f"P{i + 1}" for i in range(8)]
sf = [2, 8, 16]
ori_labels = ['horizontal', 'vertical', 'minus45', 'plus45']
```

```python
"""
Analyse data from each experiment
"""

# Analyse each experiment #
for i_exp in ['Contrast Detection', 'Orientation Discrimination']:
    # Paths #
    data_dir = os.path.join(og_dir, i_exp, 'Analysis')

    # Raw data
    raw_data = pd.read_csv(os.path.join(data_dir, "raw_anonymised.csv"))

    # Mean (SE) thresholds for each participant
    indiv_data = get_mean_data(raw_data, cond_list=['participant', 'orientation', 'spatial frequency'], dv='threshold')
    indiv_data = custom_sort(indiv_data, ['spatial frequency', 'orientation'],
                            ordered_labels=[[2, 8, 16], ori_labels])
    to_csv_pkl(indiv_data, data_dir, 'indiv_thresholds', rnd=3, _pkl=False)  # Save

    # Mean (SE) thresholds across all participants
    mean_data = get_mean_data(raw_data, cond_list=['orientation', 'spatial frequency'], dv='threshold')
    mean_data = custom_sort(mean_data, ['spatial frequency', 'orientation'],
                            ordered_labels=[[2, 8, 16], ori_labels])
    to_csv_pkl(mean_data, data_dir, 'mean_thresholds', rnd=3, _pkl=False)  # Save

    # Individual and geomean ratio data comparing performance at different orientations - e.g., oblique effect index (OEI) scores #
    ratio_data, geomean_r_data = get_ratio_data(indiv_data, cond_list=['participant', 'spatial frequency'],
                                                ori_label='orientation', ratio_list=['OEI', 'HVI'], dv='mean')
    ratio_data = custom_sort(ratio_data, ['ratio', 'participant', 'spatial frequency', 'orientation'],
                             ordered_labels=[['OEI', 'HVI'], participants, [2, 8,16], ori_labels])
    geomean_r_data = custom_sort(geomean_r_data, ['ratio', 'participant', 'spatial frequency', 'orientation'],
                                 ordered_labels=[['OEI', 'HVI'], participants, [2, 8, 16], ori_labels])
    to_csv_pkl(ratio_data, data_dir, 'ratios', rnd=3, _pkl=False)
    to_csv_pkl(geomean_r_data, data_dir, 'geomean_ratios', rnd=3, _pkl=False)  # Save

    # Statistics
    # 2-way repeated measures anova
    anova_data = indiv_data.copy()
    # Filter data to include conditions of interest
    anova_data = index_dframe(anova_data, ['orientation'], ['horizontal', 'vertical', 'minus45', 'plus45'],
                              index_filter='or')
    anova_data = index_dframe(anova_data, ['spatial frequency'], [2, 8, 16], index_filter='or')
    to_csv_pkl(anova_data, data_dir, "anova_data", rnd=3, _pkl=False)

    # - Jupyter Output
    display(Markdown(f"## **Experiment: {i_exp}**"))
    display(Markdown("**Raw Data Preview:**"), raw_data.head(8))
    display(Markdown("**Individual Thresholds Preview:**"), indiv_data.head(8))
    display(Markdown("**Mean Thresholds Preview:**"), mean_data.head(8))
    display(Markdown("**Individual Ratios Preview:**"), ratio_data.head(8))
    display(Markdown("**Geometric Mean Ratios Preview:**"), geomean_r_data.head(8))
```
